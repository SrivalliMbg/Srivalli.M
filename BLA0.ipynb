## **Inner product**

an inner product is a generalized dot product that combines two vectors to produce a single number.The inner product is also known as the dot product or scalar product.


i.e., if a,b are column matrices with same dimention,

### the inner poduct of a,b= (a^T).b

## **Outer product**

 the outer product of two vectors is a matrix where each entry is the product of an element from the first vector and an element from the second vector.

i.e.,
### the outer porduct of a,b= a.(b^T)





```python
import numpy as np
```

```python
#to calculate inner product of a,b
a = np.array([3,7,5])
b = np.array([6,3,9])
inner_product = np.inner(a,b)
print(inner_product)
```

```python
#to calculate outer broduct of a,b
 a = np.array([6,8,4])
 b = np.array([9,3,7])
 outer_product = np.outer(a,b)
 print(outer_product)
```

## **Norm of matrix**

The norm of a matrix is a real number which is a measure of the magnitude of the matrix.Norm of a matrix is defined as square root of sum of squares of matrix elements.

### Norm of matrix= sqrt(trace of matrix)

```python
 a = ([6,8,4],[4,6,2],[8,4,2])
 norm=np.linalg.norm(a)
 print(norm)
```
